```python
def persegi(sisi):
    """
    Menghitung luas dan keliling persegi.
    Parameters
    ----------
    sisi : int atau float
        Panjang sisi persegi.
    Returns
    -------
    tuple
        Dua nilai berupa (luas, keliling) dari persegi.
    """
    luas = sisi * sisi
    keliling = 4 * sisi
    return luas, keliling

# Memanggil fungsi dan menyimpan hasil ke dua variabel
l,k = persegi(10)

# Mencetak hasil perhitungan
print(f"Persegi dengan panjang 10 mempunyai luas = {l} dan keliling = {k}")
```

```python
def inifungsi(x, y=50):  # Fungsi dengan parameter x (wajib) dan y (opsional, default=50)
    print("x: ", x)  # Mencetak nilai parameter x
    print("y: ", y)  # Mencetak nilai parameter y
# Memanggil fungsi hanya dengan argumen x=10, y akan menggunakan nilai default 50
inifungsi(10)
```

```python
# Fungsi dengan dua parameter: depan dan belakang
def nama(depan, belakang):
    # Mencetak hasil penggabungan string depan dan belakang
    print(f"{depan} {belakang}")

# Memanggil dengan keyword argument (urutan bebas)
nama(depan="Alex", belakang="Umam")
# Keyword argument, urutan dibalik tapi hasil sama
nama(belakang="Umam", depan="Alex")
```

```python
# Fungsi dengan dua parameter: uang (int) dan pengirim (str)
def donasi(uang, pengirim):
    print(f"{uang} rupiah dari {pengirim}")

# Pemanggilan dengan urutan benar
donasi(10000, "Alex")  # Output: 10000 rupiah dari Alex

# Pemanggilan dengan urutan terbalik (salah urutan)
donasi("Alex", 10000)  # Output: Alex rupiah dari 10000
```

```python
# Fungsi dengan args (*nama) dan kwargs (**karakteristik)
def fungsilagi(*nama, **karakteristik):
    # Mencetak nama terakhir (index -1) dan nama pertama (index 0)
    print(f"Halo {nama[-1]}, {nama[0]}")

    # Mencetak semua keyword arguments yang diberikan
    for key, val in karakteristik.items():
        print(f"{key} = {val}")

# Memanggil fungsi dengan 3 positional arguments dan 2 keyword arguments
fungsilagi("Alzimna", "Badril", "Umam", ipk=1.0, tahun_lulus=2019)
```

```python
class Mahasiswa:
    """
    Kelas untuk merepresentasikan data mahasiswa.

    Attributes
    ----------
    departemen : str
        Departemen tempat mahasiswa berada (class attribute).
    prodi : str
        Program studi mahasiswa (instance attribute).
    nama : str
        Nama mahasiswa (instance attribute).
    tahunmasuk : int
        Tahun masuk mahasiswa (instance attribute).
    """

    departemen = "Matematika"  # Class attribute (dimiliki oleh semua instance)

    def __init__(self, prodi, nama, tahunmasuk):
        """
        Inisialisasi objek Mahasiswa.
        """
        self.prodi = prodi          # Instance attribute
        self.nama = nama            # Instance attribute
        self.tahunmasuk = tahunmasuk  # Instance attribute

    def semester(self):
        """
        Menghitung jumlah semester yang sudah ditempuh.
        """
        return (2026 - self.tahunmasuk) * 2
```

```python
# Membuat instance objek Mahasiswa
prodi = "Matematika"
nama = "Alzim"
tahunmasuk = 2015
Alzim = Mahasiswa(prodi, nama, tahunmasuk)

# Memanggil method semester
print(Alzim.semester())  # Output: 22

# Mengakses atribut prodi dari objek Alzim
print(Alzim.prodi)  # Output: Matematika
```

```python
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
plt.rcParams["savefig.dpi"] = 600

# URL GitHub untuk data
GITHUB_URL = "https://github.com/alzimnara/modul_praktikum/raw/refs/heads/main/"

# Load data batas administrasi Kota Yogyakarta
url = GITHUB_URL + "data/raw/Batas%20Administrasi%20Kota%20Yogyakarta/Batas_Administrasi.shp"
batas = gpd.read_file(url)

# Load data bangunan pemerintahan Kota Yogyakarta
url = GITHUB_URL + "data/raw/Bangunan%20Pemerintahan%20Kota%20Yogyakarta/Bangunan_Pemerintahan.shp"
pemerintahan = gpd.read_file(url)

# Load data jalan Kota Yogyakarta
url = GITHUB_URL + "data/raw/Jalan%20Kota%20Yogyakarta/Jalan.shp"
jalan = gpd.read_file(url)
```

```python
f, ax = plt.subplots(figsize=(6, 6))

# Plot boundary
batas.plot(
    ax=ax,
    facecolor="none",
    edgecolor="red",
    linewidth = 2
)

# Plot boundary
kecamatan.plot(
    ax=ax,
    facecolor="none",
    edgecolor="red",
    linewidth = 1
)

# Plot boundary
kraton.plot(
    ax=ax,
    facecolor="blue",
    edgecolor="red",
    alpha = 0.5
)

# Menambahkan annotation nama kecamatan
for idx, row in kecamatan.iterrows():
    # Mendapatkan titik tengah (centroid) dari setiap kecamatan
    centroid = row.geometry.centroid
    x = centroid.x
    y = centroid.y

    # Menambahkan teks nama kecamatan
    ax.annotate(
        row['kecamatan'],  # sesuaikan dengan nama kolom
        xy=(x-0.005, y),
        xytext=(5, 5),
        textcoords='offset points',
        fontsize=8,
        fontweight='bold',
        color='black'
        )


ax.set_title('')
ax.set_axis_off()

plt.tight_layout()
plt.savefig("kecamatan.png", bbox_inches="tight")
plt.show()
```

#**Mengklasifikasikan Nilai pada Data Raster**

```python
# Mendefinisikan batas-batas kelas elevasi (dalam meter)
class_bins = [0, 80, 120, np.inf]

# Mengaplikasikan fungsi digitize dari numpy ke data DEM
dem_classified = xr.apply_ufunc(np.digitize, dem, class_bins)

# Menghapus nilai kelas 4 (tidak termasuk dalam interval)
dem_classified = dem_classified.where(dem_classified != 4)
```

```python
f, ax = plt.subplots(figsize=(6, 6))

# Define the colors you want
cmap =ListedColormap(['#66bd63', '#d9ef8b', '#8c510a'])

im = dem_classified.plot(
    ax=ax,
    cmap=cmap,
    alpha=0.9,
    add_colorbar=False
)

# Plot boundary
batas.plot(
    ax=ax,
    facecolor="none",
    edgecolor="red",
    linewidth = 2
)

# Add a legend for labels
legend_labels = {"#66bd63": "rendah",
                 "#d9ef8b": "menengah",
                 "#8c510a": "tinggi"}

patches = [Patch(color=color, label=label)
           for color, label in legend_labels.items()]

ax.legend(handles=patches,
          bbox_to_anchor=(1.35, 1),
          facecolor="white")

# plt.legend(loc='upper right')
ax.set_title('')
ax.set_axis_off()

plt.savefig("dem_classified.png", bbox_inches="tight")
plt.tight_layout()
plt.show()
```
